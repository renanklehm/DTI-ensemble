```python
import pandas as pd
```

```python
df = pd.read_parquet("data/1_ready/df.parquet")
```

```python
df[df["label"]==df["real_value"]]
```

```python
df["y_true"] = df["real_value"].astype(int)
df["y_pred"] = df["label"].astype(int)

for source, group in df.groupby("source"):
    print(f"\nSource: {source}")

    cm = pd.crosstab(
        group["y_true"],
        group["y_pred"],
        rownames=["Actual"],
        colnames=["Predicted"],
        dropna=False
    )

    cm = cm.reindex(index=[0, 1], columns=[0, 1], fill_value=0)

    print(cm)
```
